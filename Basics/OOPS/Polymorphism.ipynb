```python
##Base class
class Animal:
    def speak(self):
        return "sound of animal"

##Derived class 1
class Dragon(Animal):
    def speak(self):
        return "Fireeeeeee!"

##drived class 2
class Cat(Animal):
    def speak(self):
        return "meow!"


##Function that demonstrates polymorphism
def animal_speak(animal):
    print(animal.speak())


dragon = Dragon()
cat = Cat()

print(dragon.speak())
print(cat.speak())
animal_speak(dragon)
```

```python
animal_speak(cat)
```

```python
##Base Class
class Shape:
    def area(self):
        return "the area of the figure"

##Derived 01
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

##Derived 02
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


##Function that demonstate polymorphism

def print_area(shape):
    print(f"The area is {shape.area()}")

rectangle = Rectangle(4,5)
circle = Circle(3)

print_area(rectangle)
print_area(circle)
    

```

```python
##Polymorphism with Abstact base class

from abc import ABC,abstractmethod

class Vehicle:
    @abstractmethod
    def start_engine(self):
        pass


class Car(Vehicle):
    def start_engine(self):
        return "The Car engine started"

class Motorcycle(Vehicle):
    def start_engine(self):
        return "The Motorcycle engine started"

def start_vehicle(vehicle):
    print(vehicle.start_engine())


car = Car()
MC = Motorcycle()

start_vehicle(car)
```

```python
start_vehicle(MC)
```
